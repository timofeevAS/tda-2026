## Построение диаграмм персистентности

После восстановления фазового пространства каждый временной ряд представлен облаком точек. Для такого облака строится фильтрация симплициальных комплексов: при увеличении масштаба соседние точки соединяются, затем появляются более высокоразмерные симплексы. На каждом масштабе отслеживаются топологические признаки.

Диаграмма персистентности содержит пары `(birth, death)`, где `birth` - масштаб появления признака, а `death` - масштаб его исчезновения. Чем дальше точка от диагонали `birth = death`, тем дольше живет соответствующий признак и тем более устойчивым он считается.
