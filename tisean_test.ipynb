```python
import pandas as pd

df = pd.read_csv("dataset_clean.csv")
```

```python
df["Lab1_G3_Lm"].to_csv("series.txt", index=False,header=False)
```

```python
import numpy as np
import plotly.graph_objects as go

# Загружаем результат lyap_r
data = np.loadtxt("series.txt.ros")

k = data[:, 0]
log_div = data[:, 1]

# Строим график Rosenstein
fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=k,
        y=log_div,
        mode="lines+markers",
        name="log divergence"
    )
)

fig.update_layout(
    title="Rosenstein curve from TISEAN lyap_r",
    xaxis_title="k",
    yaxis_title="average log divergence",
    width=900,
    height=600
)

fig.show()
```

```python
data = np.loadtxt("series.txt.ros")

k = data[:, 0]
log_div = data[:, 1]

# Ручной выбор линейного участка
k_min = 0
k_max = 1.5

mask = (k >= k_min) & (k <= k_max)

# Линейная аппроксимация: y = lambda * k + b
lambda_est, b = np.polyfit(k[mask], log_div[mask], 1)

print("Lyapunov exponent:", lambda_est)

# Линия аппроксимации
fit_y = lambda_est * k[mask] + b

fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=k,
        y=log_div,
        mode="lines+markers",
        name="TISEAN lyap_r output"
    )
)

fig.add_trace(
    go.Scatter(
        x=k[mask],
        y=fit_y,
        mode="lines",
        name=f"linear fit, λ = {lambda_est:.5f}"
    )
)

fig.update_layout(
    title=f"Lyapunov exponent estimation: λ = {lambda_est:.5f}",
    xaxis_title="k",
    yaxis_title="average log divergence",
    width=900,
    height=600
)

fig.show()
```
